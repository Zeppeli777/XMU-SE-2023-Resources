### 题目1：数组分块求和
随机创建一个形状为 (4, 4) 的整数数组（元素范围0-100）。将其划分为 2x2 的子数组，并计算每个子数组的和。



```python
# 设置随机种子以便结果可重现
np.random.seed(42)

arr = np.random.randint(0, 101, (4, 4))
print(arr)

# reshape成 (2,2,2,2)，再按块求和
sub_sum = arr.reshape(2, 2, 2, 2).sum(axis=(2, 3))
print(sub_sum)
```

### 题目2：数组滑动窗口
随机生成一个长度为10的一维整数数组（元素范围0-100）。计算大小为 3 的滑动窗口的平均值（取整）。


```python
# 设置随机种子以便结果可重现
np.random.seed(42)

arr = np.random.randint(0, 101, 10)
print("原始数组:", arr)

# 滑动窗口均值
res = np.convolve(arr, np.ones(3, dtype=int), 'valid') // 3
print("滑动窗口平均值(取整):", res)
```

### 题目3：数组对角线填充
给定一个形状为 (5, 5) 的数组，将其主对角线元素填充为 1，对角线两侧的元素填充为 2，其余元素为 0。

```python
n = 5
mat = np.zeros((n, n))

# 主对角线
np.fill_diagonal(mat, 1)

# 上下对角线
for i in range(n):
    if i + 1 < n:
        mat[i, i+1] = 2
        mat[i+1, i] = 2

print(mat)
```

### 题目4：数组条件替换
随机生成一个长度为10的整数数组（元素范围0-9），将所有大于 5 的元素替换为 1，其余替换为 0。

```python
# 设置随机种子以便结果可重现
np.random.seed(42)

arr = np.random.randint(0, 10, 10)
print("原始数组:", arr)

res = np.where(arr > 5, 1, 0)
print("替换后数组:", res)
```

### 题目5：数组分组统计
随机生成一个长度为9的整数数组（元素范围0-9），随机生成一个长度为9的分组索引，索引取值为0，1，2。求每组的平均值


```python
# 设置随机种子以便结果可重现
np.random.seed(22)

arr = np.random.randint(0, 10, 9)
groups = np.random.randint(0, 3, 9)
print("原始数组:", arr)
print("分组索引:", groups)

res = [arr[groups == i].mean() for i in range(3)]
print(res)
```

### 题目6：数组元素筛选

### 题目7：数组统计计算
创建一个4×5的随机整数数组（元素范围10-50），计算：
- 所有元素的平均值、中位数
- 每列的最大值、每行的最小值
- 数组的标准差、方差
- 数组元素总和

```python
# 设置随机种子以便结果可重现
np.random.seed(42)

arr = np.random.randint(10, 51, (4, 5))
print("随机数组:\n", arr)

print("平均值: %.2f" % arr.mean())
print("中位数: %.2f" % np.median(arr))
print("每列最大值:", arr.max(axis=0))
print("每行最小值:", arr.min(axis=1))
print("标准差: %.2f" % arr.std())
print("方差: %.2f" % arr.var())
print("元素总和:", arr.sum())
```

### 题目8：数组归一化
随机生成一个长度为10的整数数组（元素范围0-9），将其归一化到 [0, 1] 范围，小数点后保留2位。


```python
# 设置随机种子保证结果可重现
np.random.seed(42)

arr = np.random.randint(0, 10, 10)
print("原始随机数组:", arr)

res = (arr - arr.min()) / (arr.max() - arr.min())
print("归一化后的数组:", np.round(res, 2))
```

### 题目9：数组逻辑运算
给定两个布尔数组 a = [True, False, True] 和 b = [False, False, True]，计算它们的逻辑与、逻辑或和数组a的逻辑非。


```python
a = np.array([True, False, True])
b = np.array([False, False, True])

print("逻辑与:", np.logical_and(a, b))
print("逻辑或:", np.logical_or(a, b))
print("逻辑非:", np.logical_not(a))
```

### 题目10：数组拼接与拆分
给定两个数组 a = [1, 2, 3, 4] 和 b = [5, 6, 7, 8]，将它们按行拼接成一个新数组，然后将新数组再水平拆分（按列拆分）为两个子数组。


```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

c = np.vstack((a, b))
print("拼接后数组:\n", c)

arr1, arr2 = np.hsplit(c, 2)
print("拆分后数组1:\n", arr1)
print("拆分后数组2:\n", arr2)
```

### 题目11：数组标准化
随机生成一个长度为10的整数数组（元素范围10-50），将其z-score标准化，计算标准化后的数据的均值和标准差，小数点后保留2位。。


```python

# 设置随机种子保证结果可重现
np.random.seed(42)

arr = np.random.randint(10, 51, 10)
print("原始数组:", arr)

mean = arr.mean()
std = arr.std()
z = (arr - mean) / std

print("标准化后的数组:", np.round(z, 2))
print("标准化后数据的均值: %.2f" % z.mean())
print("标准化后数据的标准差: %.2f" % z.std())
```
